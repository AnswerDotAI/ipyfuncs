# ipyfuncs

> Kernel-side InteractiveShell services for kernel clients

Importing `ipyfuncs` patches `InteractiveShell` with services a kernel client can call remotely: `user_items`, `get_vars`, `eval_exprs`, `get_schemas`, `publish`, `ranked_complete`, and `sig_help`. There is no extension machinery to load -- the import is the whole setup. It was built as the kernel-side half of `conkernelclient`'s `_ipy_funcs` battery, and needs only `fastcore` and `ipython`.


## Usage

### Installation

Install latest from [pypi](https://pypi.org/project/ipyfuncs/):

```sh
$ pip install ipyfuncs
```
